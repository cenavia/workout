## Import the Agent

```python
from agents import agent

# Display the agent graph
agent.get_graph().print_ascii()
```

## Test Basic Conversation

```python
from langchain_core.messages import HumanMessage

# Test a greeting
result = agent.invoke({
    "messages": [HumanMessage(content="Hi! I'm new to fitness and want to get started.")]
})

print(result["messages"][-1].content)
```

## Test Workout Generation

```python
# Request a workout plan
result = agent.invoke({
    "messages": [HumanMessage(content="Create a 30 minute beginner full body workout. I only have dumbbells at home.")]
})

print(result["messages"][-1].content)
```

```python
# Check the extracted profile
print("\n📊 Extracted User Profile:")
print(result.get("user_profile", {}))
```

## Test Exercise Explanation

```python
# Ask about an exercise
result = agent.invoke({
    "messages": [HumanMessage(content="How do I do a proper squat? What are common mistakes?")]
})

print(result["messages"][-1].content)
```

## Test RAG Retrieval

```python
# Ask a question that should use RAG
result = agent.invoke({
    "messages": [HumanMessage(content="What's the difference between HIIT and LISS cardio?")]
})

print(result["messages"][-1].content)
print("\n📚 Retrieved Documents:")
for doc in result.get("retrieved_docs", []):
    print(f"- {doc.page_content[:100]}...")
```
